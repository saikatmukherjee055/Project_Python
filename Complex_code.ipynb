```python
import pandas as pd

# Load the data into a DataFrame
data = {
    "Name": ["Alice", "Bob", None, "David"],
    "Age": [25, 30, 29, None],
    "Email": ["alice@example.com", None, "bob@example.com", "david@example.com"]
}
df = pd.DataFrame(data)

# Check for missing values
print("Missing Values:\n", df.isnull().sum())

# Drop duplicates (if any)
df = df.drop_duplicates()

# Fill missing values with defaults
df['Name'] = df['Name'].fillna('Unknown')
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Email'] = df['Email'].fillna('no_email@example.com')

df
```

```python
data = {'Value': [10, 15, 14, 12, 100, 13, 11]}
df = pd.DataFrame(data)
df
```

```python
Q1 = df['Value'].quantile(0.25)
print(Q1)
Q3 = df['Value'].quantile(0.75)
print(Q3)
IQR = Q3 - Q1
print(IQR)

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df['Outlier'] = (df['Value'] < lower_bound) | (df['Value'] > upper_bound)
print(df)

```

```python
from functools import lru_cache

@lru_cache
def fibo(x):
    if x <=0:
        return 0
    elif x==1:
        return 1
    else:
        return fibo(x-1)+fibo(x-2)
print([fibo(x) for x in range(150)])
```

```python
class Metaclass_self(type):
    def __new__(cls,name,bases,dct):
        if 'cool_func' not in dct:
            raise TypeError(f"cool_func must be included in class")
        return super().__new__(cls,name,bases,dct)

class New_class(metaclass=Metaclass_self):
    def __init__(self):
        print("Hello")
    def cool_func():
        pass

New_class()


```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Function is called")
        return func(*args, **kwargs)
    return wrapper

@decorator
def greet(name):
    return f"Hello, {name}"

print(greet("Alice"))
```

```python
import cProfile

def slow_function():
    total = 0
    for i in range(1, 10000):
        total += sum(range(i))
    return total



# Profile slow_function
cProfile.run('slow_function()')
```

```python
import cProfile
import pstats

# Save results
with open('profile_results.prof', 'w') as f:
    profiler = cProfile.Profile()
    profiler.run('slow_function()')
    stats = pstats.Stats(profiler, stream=f)
    stats.strip_dirs()
    stats.sort_stats('time')
    stats.print_stats()
```
